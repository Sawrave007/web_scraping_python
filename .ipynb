```python
col_data = table.find_all('tr')
```

```python

num_cols = len(df)

# Tracks {column_index: (remaining_rowspan_count, value)}
active_rowspans = {}

rows_out = []
for row in col_data[2:]:
    cells = row.find_all('td')
    row_values = [None] * num_cols
    col_idx = 0

    for cell in cells:
        # skip columns currently filled by an active rowspan from above
        while col_idx in active_rowspans and active_rowspans[col_idx][0] > 0:
            row_values[col_idx] = active_rowspans[col_idx][1]
            active_rowspans[col_idx] = (active_rowspans[col_idx][0] - 1, active_rowspans[col_idx][1])
            col_idx += 1

        text = cell.text.strip()
        row_values[col_idx] = text

        rowspan = int(cell.get('rowspan', 1))
        if rowspan > 1:
            active_rowspans[col_idx] = (rowspan - 1, text)

        col_idx += 1

    # fill any trailing rowspan columns after the loop
    while col_idx < num_cols:
        if col_idx in active_rowspans and active_rowspans[col_idx][0] > 0:
            row_values[col_idx] = active_rowspans[col_idx][1]
            active_rowspans[col_idx] = (active_rowspans[col_idx][0] - 1, active_rowspans[col_idx][1])
        col_idx += 1

    rows_out.append(row_values)

df = pd.DataFrame(rows_out, columns= title_table)
```
